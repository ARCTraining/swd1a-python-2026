# Part 1 Consolidation Exercise: Modelling Your Own Data

Throughout Part 1 of this course (covering modules 01-06) we've worked with survey response data. For this consolidation exercise, think about a type of data that is relevant to your own research or professional practice.

Examples might include:

- Publications
- Protein sequences
- Laboratory samples
- Experimental results
- Historical records
- Survey responses

## Step 1: Define a small dataset

+ Create a list containing at least five items.
+ Each item should be represented as a dictionary containing attributes that are meaningful for your domain.

## Step 2: Write a custom function

+ Write a function that performs a simple analysis of your dataset.
+ Some examples might include:
    - Counting how many items belong to a particular category
    - Counting how many records contain a particular attribute
    - Finding all records associated with a particular person, species, or location
    - Counting how many records meet some condition

## Step 3: Test your function

+ Call your function using your dummy dataset and check that the result matches what you would expect.
+ Think about how your function might behave with unusual inputs...
    - What happens if the dataset has missing or incomplete entries?
    - What if no matching records are found?
