# 01: Getting Started

Throughout this course we'll work through a simple research scenario.

> You are a researcher who has collected 500 responses to the question:
>
> *"What is the biggest challenge you face in your research work?"*
>
> You would like to analyse these responses and produce some simple summaries and visualisations.

Before we start writing code, let's think about how we might approach this problem.

## Step 1: Understand the problem

When starting any programming project, there's always thinking to do before writing code.

In this scenario, we might start by considering the following questions:

+ What data do I have?
+ What would I like to learn from it?
+ What tasks might be repetitive or time-consuming?
+ Could programming help me complete those tasks more efficiently?

### Exercise

Spend a minute thinking about what kind of questions this data could help answer, and which parts of the analysis could be automated using Python. For example, reading each response one-by-one to identify the most common topics could be very time-consuming.

## Step 2: Choose a development environment
 
Now that you've decided programming could help, you'll need somewhere to write and run your code.
 
There are many tools available for writing and running Python programs. Some run locally on your computer, while others run in the cloud and are accessed through a web browser.
 
Common examples include:
 
+ Terminal or Shell (local)
+ Visual Studio Code (local)
+ Jupyter Notebook (local or cloud-based)
+ Google Colab (cloud-based)
 
For this course we'll use **Google Colab**.
 
Google Colab runs entirely in your web browser and already includes Python and many commonly used scientific libraries. This allows us to start programming immediately without installing additional software.
 
The course materials are provided as `.ipynb` (Jupyter Notebook) files. This format is supported by Google Colab, Jupyter Notebook, and Visual Studio Code, making it easy to continue using the course materials in your preferred environment after the training.
