# 04: Functions

A function is a reference to a predefined block of code which performs a specific repeatable task.

We can recognise a function because it has round parentheses (brackets) after its name.

For example,

```python
str(129)
```

## Understanding Functions

We've already used a number of built-in functions provided by Python. 

For example: 

```python
print("Learning python is fun!")
```

In this example:

+ `print` is the function name.
+ `"Learning python is fun!"` is an input argument.
+ The parentheses `()` are used to pass arguments to the function.

Some functions perform an action without returning a value, while others generate a result that can be stored in a variable.

For example:

```python
response_count = len(responses)
```

Here:
+ `len` is the function name.
+ `responses` is the input argument.
+ The function **returns** a numeric value.
+ The returned value is assigned to the variable `response_count`.

## Why Write Our Own Functions?

As our programs become larger, we often find ourselves repeating the same sequence of steps. 

Rather than copying and pasting code, we can define our own functions and reuse them wherever they are needed.

Functions can help make our code: 
- Easier to read
- Easier to maintain
- Easier to test
- Less repetitive

Before we can use a function, we must **define** it.

We can then **call** (or invoke) the function by name, passing any required arguments.

Let's see an example...

````python
def count_word(text, word):
    """Count the number of times a word appears in a piece of text."""
    return text.count(word)

response = "Finding time for research and writing papers is challenging"

count = count_word(response, "time")

print(count)
````

## Functions vs Methods

A method is similar to a function, but it belongs to a particular type of object and is called using **dot syntax**.

For example:

```python
text = "Python in Fun"

text.lower()
```

In this example:

- `text` is a string object.
- `lower()` is a method associated with strings.
- The dot (`.`) is used to access the method.

The result of the expression above is:

```python
"python is fun"
```

Methods can accept arguments in the same way as functions:

```python
text.replace("Fun", "Useful")
```

which returns:

```python
"Python is Useful"
```

Many Python objects have their own methods. For example, strings, lists, and dictionaries all provide methods that can help us manipulate and analyse data.
