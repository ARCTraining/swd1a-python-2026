Data structures allow us to group related values together in a single variable.

For example, rather than storing every survey response as an individual variable, we could group them together under a single variable.

We'll start by exploring Python's built-in data structures. Later, we'll encounter additional data structures provided by external libraries that are commonly used for scientific computing and data analysis.

## Lists

+ Lists are collections of values stored under a single variable name.
+ Lists are defined using square brackets ([]) with items separated by commas.
+ List items can be of any type, including numbers, text, or even other lists.
+ Lists can contain a mixture of different types, although using similar types is often easier to work with.
+ Lists are ordered, meaning items remain in a defined sequence.
+ Lists are mutable, meaning items can be added, removed, or modified after the list is created.
