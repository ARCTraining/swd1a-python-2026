# 01: Getting Started

Throughout this course we'll work through a simple research scenario.

> You are a researcher who has collected 500 responses to the question:
>
> *"What is the biggest challenge you face in your research work?"*
>
> You would like to analyse these responses and produce some simple summaries and visualisations.

Before we start writing code, let's think about how we might approach this problem.

## Step 1: Understand the problem

When starting any programming project, there's always thinking to do before writing code.

In this scenario, we might start by considering the following questions:

+ What data do I have?
+ What would I like to learn from it?
+ What tasks might be repetitive or time-consuming?
+ Could programming help me complete those tasks more efficiently?

### Exercise: plan your analysis

Spend a minute thinking about what kind of questions this data could help answer, and which parts of the analysis could be automated using Python. For example, reading each response one-by-one to identify the most common topics could be very time-consuming.

## Step 2: Choose a development environment
 
Now that you've decided programming could help, you'll need somewhere to write and run your code.
 
There are many tools available for writing and running Python programs. Some run locally on your computer, while others run in the cloud and are accessed through a web browser.
 
Common examples include:
 
+ Terminal or Shell (local)
+ Visual Studio Code (local)
+ Jupyter Notebook (local or cloud-based)
+ Google Colab (cloud-based)

For this course we'll use [Google Colab](colab.research.google.com).

Google Colab runs entirely in your web browser and already includes Python and many commonly used scientific libraries. This allows us to start programming immediately without installing additional software.
 
The course materials are provided as `.ipynb` (Jupyter Notebook) files. This format is supported by Google Colab, Jupyter Notebook, and Visual Studio Code, making it easy to continue using the course materials in your preferred environment after the training.

## Step 3: Take a tour of Google Colab

A Colab notebook is made up of cells.

There are two main types:
+ **Text cells**, which contain explanations and notes written in markdown.
+ **Code cells**, which contain Python code.

Each cell can be run independently, making it easy to experiment with code, explore data, and document your analysis in an interactive, shareable notebook.

### Exercise: explore Google Colab

Take a few minutes to:

+ Create a new notebook, giving it a meaningful name
+ Locate the "Run" button beside a code cell
+ Run a simple code cell (example code: `print("Hello World!")`). 
    - *Where does the output appear?*
+ Create a new text cell
+ Create a new code cell
+ Run some error-prone code (example: `print(Hello World!)`). 
    - *What happens?*
    - *How would you explain this error?*
+ Locate the "Run all" button from the top menu bar to run all your cells as a batch



## Step 4: Think about the data and code
 
For this course, you'll be given some pre-prepared dummy data to work with.
 
In a real project, there are a few additional questions you may need to consider before you begin analysing data with Python:
 
+ Where will your data be stored while it is being processed?
+ How will data and code be transferred between different systems?
+ How will your analysis and results be backed up?
+ Who needs access to your code and data?
+ How will you keep track of changes to your code as your analysis evolves?
+ How will you share your code so that others can understand, reproduce, or build upon your work?
 
Some of these considerations may seem familiar from your previous experiences with data management planning, while others may seem new.
 
We'll return to many of these topics in **SWD3: Software Development Practices for Research**, but they're worth keeping in mind from the start, as you'll often encounter them as soon as you begin using code in your research.
