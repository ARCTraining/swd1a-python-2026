# 02: Variables and assignment

> *What is a computer program?*

A computer program is a set of instructions that tells a computer how to perform a task.
 
To be useful, those instructions often need to work with information that may change as the program runs. For example:
 
- A set of survey responses read from a file
- The result of a calculation
- The number of responses in the data set
 
To keep track of this information, we use **variables**.
 
A variable is simply a name that refers to a value. By giving values meaningful names, we can reuse them later, update them, and combine them with other values as part of a larger calculation or analysis. The practice of defining variables is known as **assignment.**

Let's practice assigning values to variables in our survey scenario...

```python
# define a variable pointing to a numeric value
n_participants = 500

# define a variable pointing to a text value
response_text = "Finding time to work on it around teaching"

# define a variable pointing to a floating point (decimal) value
sentiment_score = 0.7

# define a variable referencing a boolean (true/false) value
positive_sentiment = True
```

In the above example, notice that...

+ The names given to variables are meaningful. *Why do you think this is important?*
+ The lines prefixed with `#` are ignored by the interpreter. *What do you think they are there for?*
