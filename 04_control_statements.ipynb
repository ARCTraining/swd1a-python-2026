# 04: Control Statements

+ **Control statements** allow us to control the flow of a program.
+ They allow our code to make decisions or repeat instructions based on some **condition**.
+ Two common control statements are **if statements** and **for loops**.
+ In Python, **indentation** is used to show which lines of code belong to a particular control statement.
+ Code with incorrect indentation will often produce an error or behave differently from what you intended.

## If statements
+ If statements are used to test whether a condition is true before executing a block of code.
+ They are written in the form `if condition:` and are also known as **conditional statements**.
+ They can be combined with `elif` (else if) and `else` statements to choose between multiple possible actions.
+ The condition is evaluated using **Boolean** (`True`/`False`) logic and can include comparison operators such as `==`, `!=`, `<`, and `>`, and keyword operators such as `is`, `in` and `not`.
+ Multiple conditions can be combined using the `and` and `or` keywords.
  - `and` requires all conditions to be true.
  - `or` requires at least one condition to be true.

For more on conditional statements, refer to:
+ [https://docs.python.org/3/tutorial/controlflow.html#if-statements](https://docs.python.org/3/tutorial/controlflow.html#if-statements)
+ [https://www.geeksforgeeks.org/python/conditional-statements-in-python/](https://www.geeksforgeeks.org/python/conditional-statements-in-python/)

````python
response = {
    "id": 1,
    "faculty": "Sociology",
    "role": "Research Associate",
    "text": "pressure to publish"
}

# use an if statement test a condition
if "id" in response:
    print(response["id"])
    
# optionally include an else clause
else:
    print("no id in response data")

# use an if statement to check a value exists
if response["id"]:
    print("id exists")
````

## For loops

+ **For loops** allow us to repeat a block of code a certain number of times
+ Often they are used to **iterate** (loop) through a collection of things, such as a list

````python
# for loop over a list
for item in response:
    print(item)

# for loop over indices
for i in range(1,11):
    print("participant " + str(i))
````
